Antes de sumergirnos en los algoritmos y los datos, el primer paso es familiarizarnos con nuestra herramienta de trabajo: Jupyter Notebooks (o Google Colab).

¿Cómo funciona un Notebook?
Un notebook se compone de bloques individuales llamados celdas. Existen dos tipos principales de celdas:

- Celdas de Texto (Markdown): Como esta que estás leyendo ahora. Sirven para documentar, explicar conceptos y estructurar nuestro proyecto.
- Celdas de Código (Python): Celdas interactivas donde escribiremos y ejecutaremos nuestro código.
