# Informe de ventas Q3

Análisis del margen medio por región.
Resultado final: **margen medio del 41,8 %**

```python
import numpy as np
import pandas as pd
```

```python
ventas = np.array([1200., 950., 1430., 880., 1610.])
costes = np.array([700., 610., 800., 520., 990.])
```

```python
# descuento comercial aplicado a las ventas
ventas = ventas * 0.95
```

```python
margen = (ventas - costes) / ventas
margen.mean()
```

### Corrección posterior
Se detectó un coste logístico no imputado.

```python
costes = costes + 40
```

```python
print(f'Margen medio: {margen.mean()*100:.1f} %')
```

## Conclusión

El margen medio del trimestre es del 41,8 %, por encima del objetivo del 38 %.
